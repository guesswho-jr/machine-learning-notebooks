```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
dataset = "1) iris.csv"
```

```python
df = pd.read_csv(dataset)
```

```python
df.head()
```

```python
labels=df.pop("species")
features = df
```

```python
features.head(), labels.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.3)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
model = make_pipeline(StandardScaler(), KNeighborsClassifier())
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

print(f"Accuracy: {accuracy_score(y_pred=y_pred, y_true=y_test)}")
```

```python
from sklearn.metrics import confusion_matrix

print(f"Confusion Matrix: {confusion_matrix(y_pred=y_pred, y_true=y_test)}")
```

```python
from sklearn.metrics import precision_score

print(f"Precision score: {precision_score(y_pred=y_pred, y_true=y_test, average="macro")}")
```

```python
from sklearn.metrics import recall_score

print(f"Recall score: {recall_score(y_pred=y_pred, y_true=y_test, average="macro")}")
```

```python
model_2 = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=10))
```

```python
model_2.fit(X_train, y_train)
```

```python
y_pred_2 = model_2.predict(X_test)
```

```python
print("Accuracy: ", accuracy_score(y_pred=y_pred_2, y_true=y_test))
print("Confusion matrix: ", confusion_matrix(y_pred=y_pred_2, y_true=y_test))
print("Precision score: ", precision_score(y_pred=y_pred_2, y_true=y_test, average="macro"))
print("Recall score: ", recall_score(y_pred=y_pred_2, y_true=y_test, average="macro"))
```
